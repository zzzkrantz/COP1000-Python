```python
# exercise 7
def calculate_discount(price, discount_percent):
    discount_amount = price * (discount_percent / 100)
    return price - discount_amount

tests = [
    {"price": 100, "discount": 20, "expected": 80.00},
    {"price": 50, "discount": 10, "expected": 45.00},
    {"price": 25, "discount": 0, "expected": 25.00}
]

for i, t in enumerate(tests, 1):
    actual = calculate_discount(t["price"], t["discount"])
    print(f"Test {i}: Price ${t['price']}, Discount {t['discount']}% -> Expected: ${t['expected']:.2f}, Actual: ${actual:.2f}")
```
