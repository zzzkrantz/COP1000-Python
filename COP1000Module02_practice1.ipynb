```python
#creating variables
first_name = "Zach"
age = 24
gpa = 4.0
is_student = True

print(first_name)
print(age)
print(gpa)
print(is_student)

```

```python
#identifying data types
first_name = "Zach"
age = 24
gpa = 4.0
is_student = True

print(type(first_name))
print(type(age))
print(type(gpa))
print(type(is_student))
```

```python
#changing a variable
score = 75

print(score)

score = 90

print(score)
```

```python
#math with variables
number1 = 20
number2 = 5

print(number1 + number2)
print(number1 - number2)
print(number1 * number2)
print(number1 / number2)
```
