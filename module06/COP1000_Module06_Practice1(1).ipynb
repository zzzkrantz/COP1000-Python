```python
# exercise 5
names = ["Trump", "Vance", "Biden", "Harris", "Mitch McConnel"]

print("Current or former representatives include: ", names)
names.sort()
print("Current or former representatives include: ", names)
names.sort(reverse=True)
print("Current or former representatives include: ", names)
```

```python
# exercise 6
scores = [88, 94, 76, 91, 85]

print("Highest score:", max(scores))
print("Lowest score:", min(scores))

total_scores = sum(scores)
print("Total scores:", total_scores)

num_scores = len(scores)
print("Number of scores:", num_scores)

average = total_scores / num_scores
print(f"Average score: {average:.2f}")
```

```python
# exercise 7
students = []

# Ask the user to enter three student names and append to the list
for i in range(3):
    name = input(f"Enter student name {i+1}: ")
    students.append(name)

print("\nComplete list of students:", students)
print("Number of students:", len(students))

students.sort()
print("Sorted list of students:", students)
```
