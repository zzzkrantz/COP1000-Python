```python
# exercise numero tres

course = {
    "course": "COP 1000",
    "room": "A101",
    "students": 24
}
print(course)
course["room"] = "B205"
course["students"] = 26
print("Updated course:", course)
```

```python
# exercise numero cuatro

profile = {
    "name": "Taylor",
    "city": "Jacksonville",
    "age": 21,
    "favorite_color": "blue"
}
profile.pop("favorite_color", None)
print("updated profile:", profile)
print("remaining key-value pairs:", len(profile))
```

```python
# exercise numero cinco

inventory = {
    "notebooks": 12,
    "pens": 30,
    "folders": 8
}
if "pens" in inventory:
    print("We got pens.")
else:
    print("We do not got pens.")
if "markers" in inventory:
    print("We got markers.")
else:
    print("We do not got markers.")
```

```python
# exercise numero siete

contact = {}
contact["Name"] = input("Enter Name: ")
contact["Phone"] = input("Enter Phone number: ")
contact["Email"] = input("Enter Email address: ")
contact["City"] = input("Enter City: ")

print("\nContact Information\n")
for key, value in contact.items():
    print(f"{key}: {value}")
```
