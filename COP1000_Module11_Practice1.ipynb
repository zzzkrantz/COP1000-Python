```python
import math

radius = float(input('Enter the radius of the circle: '))

area = math.pi * radius**2
circumference = 2 * math.pi * radius

print(f"Area of the circle: {area:.2f}")
print(f"Circumference of the circle: {circumference:.2f}")
```

```python
import random

try:
    min_num = int(input("Enter a minimum number: "))
    max_num = int(input("Enter a maximum number: "))
    if min_num > max_num:
        print("The minimum number cannot be greater than the maximum number.")
    else:
        random_result = random.randint(min_num, max_num)
        print(f"Your random number between {min_num} and {max_num} is: {random_result}")
except ValueError:
    print("Please enter valid integers.")
```
