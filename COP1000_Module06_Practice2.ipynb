```python
temperatures = [45, 72, 88, 61, 93, 55, 79]
count = 0

print("Temperatures 70 or higher:")
for temp in temperatures:
    if temp >= 70:
        print(temp)
        count += 1

print(f"\nNumber of temperatures 70 or higher: {count}")
```

```python
scores = []

for i in range(5):
    while True:
        try:
            score = float(input(f"Enter test score {i+1}: "))
            scores.append(score)
            break
        except ValueError:
            print("Invalid input. Please enter a number.")

print(f"\nScores: {scores}")

total = sum(scores)
print(f"Total: {total}")

average = total / len(scores)
print(f"Average: {average:.2f}")

highest = max(scores)
print(f"Highest score: {highest}")

lowest = min(scores)
print(f"Lowest score: {lowest}")
```

```python
weekend = ("Saturday", "Sunday")

print(f"Complete tuple: {weekend}")
print(f"First item: {weekend[0]}")
print(f"Second item: {weekend[1]}")
print(f"Number of items: {len(weekend)}")

print("\nDays of the weekend:")
for day in weekend:
    print(day)
```
