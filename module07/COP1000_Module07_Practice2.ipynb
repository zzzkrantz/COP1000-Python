```python
# Exercise 1
colors = {"red", "blue", "green", "yellow"}
print("Colors Set:", colors)
print("Number of items:", len(colors))
```

```python
# Exercise 2
courses = {"COP 1000", "ENC 1101", "MAT 1033"}
courses.add("SLS 1101")
courses.discard("MAT 1033")
print("Final set:", courses)
```

```python
# Exercise 3
names = ["Alex", "Jordan", "Alex", "Taylor", "Jordan", "Morgan"]
print("Original list:", names)

names_set = set(names)
print("Unique names set:", names_set)

print("Number of values in original list:", len(names))
print("Number of unique values in set:", len(names_set))
```

```python
# Exercise 4
languages = {"Python", "Java", "C++", "JavaScript"}

user_input = input("Enter a programming language: ")
if user_input in languages:
    print(f"{user_input} is in the set.")
else:
    print(f"{user_input} is not in the set.")
```

```python
# Exercise 5
student_a = {"Python", "Gaming", "Music", "Travel"}
student_b = {"Python", "Music", "Photography", "Sports"}

common_interests = student_a.intersection(student_b)
print("Common interests:", common_interests)
```

```python
# Exercise 6
student_a = {"Python", "Gaming", "Music", "Travel"}
student_b = {"Python", "Music", "Photography", "Sports"}

all_unique_interests = student_a.union(student_b)
print("All unique interests:", all_unique_interests)
```

```python
# Exercise 7
python_students = {"Alex", "Jordan", "Taylor", "Morgan"}
web_students = {"Jordan", "Morgan", "Casey", "Riley"}

both_courses = python_students.intersection(web_students)
print("Students in both courses:", both_courses)

either_course = python_students.union(web_students)
print("Students in either course:", either_course)

python_only = python_students.difference(web_students)
print("Python only:", python_only)
```
